## Openning a csv file from an URL

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

### Index_col parameter

```python
pd.read_csv('placementdata.csv',index_col=0)  # index_col removes the first column from the data frame and uses it as the index of the data frame.
```

### Header parameter

```python
pd.read_csv('placementdata.csv',header=None)  # header=None tells pandas that the first row of the CSV file does not contain column names, and it should treat all rows as data.
```

### use_cols parameter

```python
pd.read_csv('placementdata.csv',usecols =['cgpa','iq'])
```

### Squeeze parameters

```python
pd.read_csv('placementdata.csv', usecols=['cgpa']).squeeze("columns")
```

### Skiprows/nrows parameter

```python
pd.read_csv('placementdata.csv',skiprows=[0,1])  # skiprows skip the first 5 rows of the data frame.
```

```python
pd.read_csv('placementdata.csv',nrows=50)  # nrows reads only the first 50 rows of the data frame.
```

### Encoding parameter

```python
pd.read_csv('placementdata.csv',encoding='latin-1')  # encoding='latin-1' tells pandas to use the Latin-1 encoding when reading the CSV file. This is useful if the CSV file contains non-ASCII characters.
```

### skip bad lines

```python
pd.read_csv('placementdata.csv', sep=';', encoding='latin-1', on_bad_lines='skip')
```

### Dtypes parameter

```python
pd.read_csv('placementdata.csv',dtype={'placement': int}).info()
```

### convertors

```python
def rename(name):
    if name == 'New York':
        return 'NY'
    else:
        return name
```

```python
rename('New York')
```

```python
pd.read_csv('placementdata.csv',converters={'city': rename})  # converters={'city': rename} tells pandas to apply the rename function to the 'city' column when reading the CSV file. This allows you to modify the values in the 'city' column as they are read into the DataFrame.
```

### Loading a huge dataset in chunks

```python
dfs = pd.read_csv("placementdata.csv", chunksize=20)  # chunksize=20 tells pandas to read the CSV file in chunks of 20 rows at a time. This is useful for processing large CSV files that may not fit into memory all at once.
```

```python
for chunk in dfs:
    print(chunk.shape)  # chunk.shape returns the number of rows and columns in the current chunk of the DataFrame. This allows you to see how many rows and columns are in each chunk as you iterate through them.
```
